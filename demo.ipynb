# Code Four Video Search - demo

Natural-language search over 6.67 hours of body-worn camera footage (20 videos).

**The idea in one line:** cheap local indexes propose candidate moments; a
vision-language model inspects only the finalists - and is allowed to say no.
Timestamps always come from the system's own bookkeeping, never a model's memory.

All outputs below are pre-executed; any cell can be re-run live. Videos play
inline (run with `uv run jupyter lab demo.ipynb` from the repo root).

## Example query 2 (from the challenge): a suspect being handcuffed

Nobody says the word "handcuffing" out loud - this is found through the caption

## Example query 3 (from the challenge): a vehicle pulled over at night

Cross-modal: the planner turns "at night" into a scene *filter* (scene attributes
flooded retrieval when they were searchable - we measured it), and the stop itself
is found through captions and transcript.

```python
search("find all instances of a vehicle being pulled over at night")
```

## A trap query: the system is allowed to say no

There is no gunshot anywhere in this corpus (verified - the closest thing anyone
says is "I'm glad it *wasn't* shots fired"). Every configuration without
verification answers this with confident nonsense; the verifier refuses it and
shows the closest rejected candidate with its reason.

```python
search("a gunshot is fired")
```

## The numbers

24 audited queries (18 answerable, 6 no-answer traps) over the full corpus,
recommended configuration:

| Hit@1 | Hit@5 | Traps refused | Seconds/query | Cost/query |
|---|---|---|---|---|
| 0.67 | 0.83 | **6/6** | ~14 | ~$0.014 |

Precision holds on the hard cross-modal stratum (0.67, equal to the easy one).
Indexing the corpus took ~12 min of local compute per video-hour and $0.90 of
API captioning total - an all-API extraction pipeline would cost ~40x more.
